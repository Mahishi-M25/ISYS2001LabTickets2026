```python
print("=== Personal Finance Tracker v0.1 ===")
print("Week 3: Smart Budget Decisions")
print()

# User setup
name = input("Enter your name: ")
monthly_income = float(input("Enter your monthly income: $"))
emergency_fund = float(input("How much do you have saved in your emergency fund? $"))

print(f"\nWelcome to your personal finance tracker, {name}!")
print(f"Monthly income: ${monthly_income:.2f}")
print(f"Emergency fund: ${emergency_fund:.2f}")
print()

# Expense categorisation
print("\n--- Expense Entry ---")
expense_amount = float(input("Enter an expense: $"))

# Financial ratios
income_ratio = expense_amount / monthly_income
emergency_ratio = expense_amount / emergency_fund if emergency_fund > 0 else float('inf')
percentage_expense = (expense_amount / monthly_income) * 100

advice = ""# Advice placeholder
category = input("Is this item a 'necessity' or a 'luxury'? ").strip().lower()
if category == "necessity":
    advice += " Since this is a necessity, it may be worth prioritising even if it stretches your budget."
elif category == "luxury":
    advice += " As this is a luxury, consider whether it aligns with your long-term financial goals before buying."
else:
    advice += " Please enter 'necessity' or 'luxury' for the category."
print()

if percentage_expense > 50:
      expense_category = "Major purchase - think carefully"
      percent_amount = "50%"
elif percentage_expense > 25:
      expense_category = "Significant expense"
      percent_amount = "25%"
elif percentage_expense > 10:
      expense_category = "Moderate expense"
      percent_amount = "10%"
else:
      expense_category = "Minor expense - within budget"
      percent_amount = "<10%"
calculation = f"This is {expense_category} - You are at {percent_amount} of your monthly income."

# Decision logic
warning = ""
if emergency_ratio > 0.5:
    warning = "⚠️ Warning: This purchase would use more than half of your emergency fund. Avoid draining your safety net."

elif income_ratio > 0.5:
    warning = "This item costs more than half of your monthly income. It may not be reasonable."

elif income_ratio <= 0.25 and emergency_ratio <= 0.25:
    warning = "✅ This purchase is affordable, reasonable compared to your income, and safe for your emergency fund. Go ahead if you need it :)"

else:
    warning = "You can afford this item, but make sure it won’t affect your other financial priorities."


# Output results
print("--- Financial Details ---")
print()
print(f"Monthly income: ${monthly_income:.2f}")
print(f"Emergency fund: ${emergency_fund:.2f}")
print()
print(f"Income ratio : {income_ratio:.2f}")
print(f"Emergency fund ratio : {emergency_ratio:.2f}")
print()
print("---Financial Advice---")
print(f"According to the percentage calculation : {calculation}\nAccording to the categorisation : {advice}\nAccording to the ratios : {warning}")
print()
# Payment plan calculator
print("--- Payment Plan Options ---")
for months in [3, 6, 12]:
    monthly_payment = expense_amount / months
    print(f"{months}-month plan: ${monthly_payment:.2f} per month")
print()
```
