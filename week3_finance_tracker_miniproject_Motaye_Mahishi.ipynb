## Part 1: Warm-Up - Pattern Recognition (10 minutes)

### From Grades to Money
You've learned to classify test scores into letter grades. Let's apply the **exact same pattern** to classify expenses:

**Grade Classifier** (familiar):
```python

    grade = "A"

    grade = "B"

```

**Expense Classifier** (new domain, same logic):
```python

    category = "Major Expense"
elif amount >= 50:
    category = "Moderate Expense"
elif amount >= 10:
    category = "Minor Expense"
else:
    category = "Small Purchase"
```

**Your Task**: Write a complete expense classifier that:

1. Asks user for an expense amount
2. Classifies it using the ranges above
3. Prints a helpful message for each category



````python
amount = int(input("Enter an expense amount: "))

if amount >= 100:
  category = "Major Expense"
  print("Reduce your expenses! It is too much!")
elif amount >= 50:
  category = "Moderate Expense"
  print("Minimize your expenses! you are at borderline")
elif amount >= 10:
  category = "Minor Expenses"
  print("You are doing great! Save more")
else:
  category = "Small Purchase"
  print("You are the queen of saving money!")
print(f"This is {category}")
````

## Part 2: Financial Decision Logic (15 minutes)

### Budget Limit Checker
Apply your if/else skills to real budget management:

````python
print("=== Budget Limit Checker ===")

# Get user's budget and expense
monthly_budget = float(input("Enter your monthly budget: $"))
expense = float(input("Enter expense amount: $"))

# Decision logic here - you write this part!
percentage_expense = (expense / monthly_budget) * 100
# If expense > 50% of budget: "Major purchase - think carefully"
if percentage_expense > 50:
  category = "Major purchase - think carefully"
  percent_amount ="50%"
# If expense > 25% of budget: "Significant expense"
elif percentage_expense >25:
  category = "Significant expense"
  percent_amount = "25%"
# If expense > 10% of budget: "Moderate expense"
elif percentage_expense > 10:
  category = "Moderate expense"
  percent_amount = "10%"
# Otherwise: "Minor expense - within budget"
else:
  print("Minor expense - within budget")
print(f"This is {category} - You are at {percent_amount} ")
````

## Part 3: Smart Purchase Advisor (20 minutes)

### The Challenge
Build a program that helps users make smart purchase decisions by considering multiple factors.

### Requirements

Your purchase advisor should:

1. Get purchase amount and user's available money
2. Check if they can afford it (basic affordability)
3. Give different advice based on purchase size
4. Consider their financial situation

### Starter Framework

````python
print("=== Smart Purchase Advisor ===")

#Get personal information
name = (input("Enter your name:"))
print(f"Welcome to your Smart Purchase Advisor - {name}")
print()

# Get financial information
available_money = float(input("How much money do you have available? $"))
purchase_price = float(input("How much does the item cost? $"))
monthly_income = float(input("What's your monthly income? $"))
emergency_fund = float(input("How much do you have saved in your emergency fund? $"))
category = input("Is this item a 'necessity' or a 'luxury'? ").strip().lower()
print()

# Calculate ratios
affordability_ratio = purchase_price / available_money
income_ratio = purchase_price / monthly_income
emergency_ratio = purchase_price / emergency_fund if emergency_fund > 0 else float('inf')

# Decision logic
if purchase_price > available_money:
    advice = "It costs more than what you have - You cannot afford this item right now. Consider saving more before purchasing."

elif emergency_ratio > 0.5:
    advice = "⚠️ Warning: This purchase would use more than half of your emergency fund. Avoid draining your safety net."

elif affordability_ratio > 0.75:
    advice = "This purchase would use more than 75% of your available money - May drain your savings. Think carefully before buying."

elif income_ratio > 0.5:
    advice = "This item costs more than half of your monthly income. It may not be reasonable."

elif affordability_ratio <= 0.25 and income_ratio <= 0.25 and emergency_ratio <= 0.25:
    advice = "✅ This purchase is affordable, reasonable compared to your income, and safe for your emergency fund. Go ahead if you need it :)"

else:
    advice = "You can afford this item, but make sure it won’t affect your other financial priorities."

# Category-based advice
if category == "necessity":
    advice += " Since this is a necessity, it may be worth prioritizing even if it stretches your budget."
elif category == "luxury":
    advice += " As this is a luxury, consider whether it aligns with your long-term financial goals before buying."

# Payment plan calculator (simple example: 3, 6, 12 months)
print("\n--- Payment Plan Options ---")
for months in [3, 6, 12]:
    monthly_payment = purchase_price / months
    print(f"{months}-month plan: ${monthly_payment:.2f} per month")

# Output results
print("\n--- Financial Ratios ---")
print(f"Affordability ratio : {affordability_ratio:.2f}")
print(f"Income ratio : {income_ratio:.2f}")
print(f"Emergency fund ratio : {emergency_ratio:.2f}")
print()
print(f"According to the ratios : {advice}")
````

### Sample Decision Logic to Implement

- **Can afford but expensive**: "You can afford it, but it's Y% of your available money. Consider waiting."
- **Reasonable purchase**: "This seems like a reasonable purchase for your budget"
- **Small purchase**: "This is a minor expense - go for it!"
