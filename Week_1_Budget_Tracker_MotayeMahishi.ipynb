## Part 3: Building the Budget Tracker

Use AI to help you write a program that:
- Allows users to input their monthly income
- Allows users to input up to 3 expenses  
- Calculates remaining budget
- Shows a warning if budget is negative

### 💡 Pro Tip: Be specific about requirements!

```python
# TASK: Write your detailed prompt for the budget tracker
"""
My Budget Tracker Prompt:
add the following to the code
write a program in python that:

Allows users to input their monthly income
Allows users to input up to 3 expenses
Calculates remaining budget
Shows a warning if budget is negative
"""
```

### Paste and Test Your AI-Generated Budget Tracker

```python
# AI-GENERATED BUDGET TRACKER CODE
# Budget Tracker Program

# Ask for the user's name
name = input("Please enter your name: ")

# Ask for the user's monthly income
while True:
    income_str = input("Please enter your monthly income: ")
    # Clean the input string: remove currency symbols, spaces, and commas
    cleaned_income_str = income_str.replace('MUR', '').replace(',', '').strip()
    try:
        income = float(cleaned_income_str)
        break # Exit loop if conversion is successful
    except ValueError:
        print("Invalid input. Please enter a numerical value for your income (e.g., 7000 or 7000.50).")

# Personalized welcome message
print(f"\nWelcome, {name}! Let's track your budget.")

# Collect up to 3 expenses
expenses = []
for i in range(1, 4):
    while True:
        expense_str = input(f"Enter expense {i} (or press Enter to skip): ")
        if expense_str.strip() == "":
            break # Allow skipping expenses
        # Clean the input string for expenses as well
        cleaned_expense_str = expense_str.replace('MUR', '').replace(',', '').strip()
        try:
            expenses.append(float(cleaned_expense_str))
            break # Exit loop if conversion is successful
        except ValueError:
            print("Invalid input. Please enter a numerical value for the expense (e.g., 100, 50.75).")

# Calculate total expenses
total_expenses = sum(expenses)

# Calculate remaining budget
remaining_budget = income - total_expenses

# Display results
print("\n--- Budget Summary ---")
print(f"Monthly Income: {income}")
print(f"Total Expenses: {total_expenses}")
print(f"Remaining Budget: {remaining_budget}")

# Warning if budget is negative
if remaining_budget < 0:
    print("⚠️ Warning: Your budget is negative! You have overspent.")
else:
    print("✅ Great! You are within your budget.")
```

Choose ONE feature to add using AI assistance:
- 🏷️ Expense categories (food, transport, entertainment)
- 📊 Show percentage of income spent
- ✅ Input validation (no negative numbers)
- 📝 Detailed budget summary with advice

```python
# Document your enhancement
enhancement_choice = "I chose to add: Show percentage of income spent"
enhancement_prompt = """
I want to add a feature to the code that will show the percentage of income spent. Along with the warning in case of negative budget, give a warning that percentage of income spent has exceeded 100.
"""

print(f"🚀 ENHANCEMENT: {enhancement_choice}")
```

```python
# ENHANCED BUDGET TRACKER
# Budget Tracker Program with Percentage Spent

# Ask for the user's name
name = input("Please enter your name: ")

# Ask for the user's monthly income
income = float(input("Please enter your monthly income: "))

# Personalized welcome message
print(f"\nWelcome, {name}! Let's track your budget.")

# Collect up to 3 expenses
expenses = []
for i in range(1, 4):
    expense = input(f"Enter expense {i} (or press Enter to skip): ")
    if expense.strip() == "":
        break
    else:
        try:
            expenses.append(float(expense))
        except ValueError:
            print("Invalid input, please enter a number.")
            continue

# Calculate total expenses
total_expenses = sum(expenses)

# Calculate remaining budget
remaining_budget = income - total_expenses

# Calculate percentage of income spent
if income > 0:
    percent_spent = (total_expenses / income) * 100
else:
    percent_spent = 0

# Display results
print("\n--- Budget Summary ---")
print(f"Monthly Income: {income}")
print(f"Total Expenses: {total_expenses}")
print(f"Remaining Budget: {remaining_budget}")
print(f"Percentage of Income Spent: {percent_spent:.2f}%")

# Warnings
if remaining_budget < 0:
    print("⚠️ Warning: Your budget is negative! You have overspent.")
    print("⚠️ Warning: You have spent more than 100% of your income!")
else:
    print("✅ Great! You are within your budget.")
```
